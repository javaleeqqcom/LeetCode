```python
from pathlib import Path

from agents.analyze_agent import AnalyzeAgent
from agents.case_generator_agent import CaseGeneratorAgent

QUESTION = Path(
    r"Question/3660. Jump Game IX/题目.txt"
).read_text(encoding="utf-8")


STUDENT_CODE = Path(
    r"Question/3660. Jump Game IX/V5.0.py"
).read_text(encoding="utf-8")

LANGUAGE = "python"
```

```python

print("=" * 80)
print("STEP 1 : Analyze")
print("=" * 80)

analyze_agent = AnalyzeAgent()

analysis = analyze_agent.run(
    question=QUESTION,
    student_code=STUDENT_CODE,
    language=LANGUAGE,
)

print(analysis.model_dump_json(indent=2))
```

```python

print("\n")
print("=" * 80)
print("STEP 2 : Generate Case Generator")
print("=" * 80)

case_agent = CaseGeneratorAgent()

case_code = case_agent.run(
    question=QUESTION,
    student_code=STUDENT_CODE,
    language=LANGUAGE,
    analysis=analysis,
)

print(case_code)


out_path = Path(
    r"Question/3660. Jump Game IX/case_generator_ai.py"
)

out_path.write_text(case_code.content, encoding="utf-8")

print("\nSaved:", out_path)
```
